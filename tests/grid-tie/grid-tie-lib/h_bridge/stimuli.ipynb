# Design Script

Use the `DesignPoints()` class to come up with your dimensions for your circuit and finally save it in a file that is called the same as this notebook but with `*.json` extension.

```python
from opens_suite import Stimuli

stimuli = Stimuli()
stimuli["VIN"] = 30
stimuli["GND"] = 0

stimuli["en_1_i"] = stimuli.vpattern([0,"dt=10n", 1,0], vdd=3.3, tr=1e-9, tf=1e-9, period=100e-9, duty=1.0)
stimuli["en_2_i"] = stimuli.vpattern([1,0,"dt=20n", 1], vdd=3.3, tr=1e-9, tf=1e-9, period=100e-9, duty=1.0)

stimuli["P1", "P2"] = stimuli.res(100)
```

```python
# This will be picked up automatically by opens when running the design script
stimuli.save("stimuli.json")
```
