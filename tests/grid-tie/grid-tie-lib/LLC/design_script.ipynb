# Design Script

Use the `DesignPoints()` class to come up with your dimensions for your circuit and finally save it in a file that is called the same as this notebook but with `*.json` extension.

```python
from opens_suite import DesignPoints
import numpy as np

dps = DesignPoints()

dps["fres [H]"] = 100e3
dps["k"] = 2
dps["vin [V]"] = 30*2 # full bridge configuration
dps["vout [V]"] = 400 # dc output voltage
dps["n"] = dps["vout"]/dps["vin"] / 2 # we have 
dps["pout [W]"] = 250
dps["R4.R [ohm]"] = dps["vout"]**2/ dps["pout"]


dps["Rac [ohm]"] = 8*1/dps["n"]**2 * dps["vout"]**2 / np.pi**2 / dps["pout"] 

Qtarget = 0.1
Z0target = Qtarget*dps["Rac"]

dps["C1.C [F]"] = 1/(2*np.pi*dps["fres"]*Z0target)
dps["TX1.LP [F]"] = 1/((2*np.pi*dps["fres"])**2 * dps["C1.C"])
dps["L1.L [F]"] = dps["TX1.LP"]/dps["k"]

dps["TX1.LS [F]"] = dps["TX1.LP [F]"] * dps["n"]**2
dps["Z0 [ohm]"] = np.sqrt(dps["L1.L"] / dps["C1.C"] )

dps["Q0"] = dps["Z0"]/dps["Rac"]
dps
```

```python

dps = DesignPoints()

dps["vin"] = 30*2
dps["vout"] = 400
dps["pout"] = 1e3
dps["fres"] = 100e3
dps["k"] = 3
dps["Qtarget"] = 0.4

# 1. Windungsverhältnis (Primär zu Sekundär)
dps["n"] = dps["vin"] / (dps["vout"]) # Halbbrücke: Vin/2 wird transformiert


dps["R4.R"] = dps["vout"]**2 / (dps["pout"])

# 2. Lastwiderstand primärseitig (Rac)
# Formel: (8/pi^2) * n^2 * R_load
R_load = dps["vout"]**2 / dps["pout"]
dps["Rac"] = (8 * dps["n"]**2 / np.pi**2) * R_load

# 3. Schwingkreis
Z0 = dps["Qtarget"] * dps["Rac"]
dps["C1.C"] = 1 / (2 * np.pi * dps["fres"] * Z0)
dps["L1.L"] = Z0 / (2 * np.pi * dps["fres"]) # Das ist Lr

# 4. Trafo
dps["TX1.LP"] = dps["k"] * dps["L1.L"] # Das ist Lm

dps["TX1.LS [F]"] = dps["TX1.LP [F]"] * (1/dps["n"])**2

dps
```

```python
# This will be picked up automatically by opens when running the design script
dps.save("design_script.json", id=0)
```
