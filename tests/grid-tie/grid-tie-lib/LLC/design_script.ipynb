# Design Script

Use the `DesignPoints()` class to come up with your dimensions for your circuit and finally save it in a file that is called the same as this notebook but with `*.json` extension.

```python
from opens_suite import DesignPoints
import numpy as np
dps = DesignPoints(["variables.json"])

dps["vin"] = dps["VIN"]
dps["vout"] = 400
dps["pout"] = 250
dps["fres"] = 100e3
dps["k"] = 10
dps["Qtarget"] = 0.3

# 1. Windungsverhältnis (Primär zu Sekundär)
dps["n"] = dps["vin"] / (dps["vout"]/2) # Halbbrücke: Vin/2 wird transformiert


dps["R4.R [Ohm]"] = dps["vout"]**2 / (dps["pout"])

# 2. Lastwiderstand primärseitig (Rac)
dps["Rac [Ohm]"] = (2 * dps["n"]**2 / np.pi**2) * dps["R4.R"]

# 3. Schwingkreis
Z0 = dps["Qtarget"] * dps["Rac"]
dps["C1.C [F]"] = 1 / (2 * np.pi * dps["fres"] * Z0)
dps["L1.L [H]"] = Z0 / (2 * np.pi * dps["fres"]) # Das ist Lr

# 4. Trafo
dps["TX1.LP [H]"] = dps["k"] * dps["L1.L"] # Das ist Lm
dps["TX1.LS [H]"] = dps["TX1.LP"] * (1/dps["n"])**2


dps["Vref [Ohm]"] = 1.1
dps["I_meas_max [A]"] = 0.1/400
dps["R6.R"] = dps["Vref"] / dps["I_meas_max"]

dps["R5.R"] = (dps["vout"]-dps["Vref"]) / dps["I_meas_max"]
dps
```

```python
# This will be picked up automatically by opens when running the design script
dps.save("design_script.json", id=0)
```
