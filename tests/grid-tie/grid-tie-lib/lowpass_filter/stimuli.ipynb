# Stimuli Generator

Use the `Stimuli()` class to come up with the stimuli and finally save it in a file that is called the same as this notebook but with `*.json` extension.

```python
from opens_suite import Stimuli

stimuli = Stimuli()

stimuli["VDD"] = 3.3
stimuli["VSS"] = 0

stimuli["vin_ai"] = stimuli.vdc(1.65, ac=1)
```

```python
# This will be picked up automatically by opens when running the design script
stimuli.save("stimuli.json")
```
