# Sallen-Key Lowpass Filter Design Script

Calculate R and C component values for a unity-gain Sallen-Key (VCVS) lowpass filter.
For a Butterworth response with equal capacitors:
- f_c = 1 / (2 * pi * R * C)
- Both capacitors should be equal for optimal response

```python
from opens_suite import DesignPoints
import math

# Design parameters
f_3db = 50  # Desired 3dB cutoff frequency (Hz) - CHANGE THIS VALUE
C = 100e-9   # Capacitor value (100nF) - use same value for both C1 and C2

# For equal-component Sallen-Key Butterworth filter:
# f_c = 1 / (2 * pi * R * C)
# Therefore: R = 1 / (2 * pi * f_c * C)
R = 1 / (2 * math.pi * f_3db * C)

dps = DesignPoints()

# Set resistor values (R2 and R3)
dps["R2.R [Ohm]"] = R
dps["R3.R [Ohm]"] = R

# Set capacitor values (C1 and C2) - equal for Butterworth response
dps["C1.C [F]"] = C
dps["C2.C [F]"] = C

print(f"Sallen-Key Butterworth Lowpass Filter Design")
print(f"="*50)
print(f"Target cutoff frequency (f_3db): {f_3db} Hz")
print(f"Capacitor values (C1, C2): {C*1e9:.1f} nF")
print(f"Resistor values (R2, R3): {R:.2f} Ω = {R/1e3:.3f} kΩ")
print(f"="*50)

dps
```
